```python
import torch
import math
```

```python
from cmath import sqrt
Q = torch.tensor([[1., 1.],
                [0., 1.],
                [1., 2.]])

K = torch.tensor([[1., 0.],
                [0., 1.],
                [1., 1.]])

K_T = K.T

S = Q@K_T

d_k = Q.shape[1]
S_scaled = S / math.sqrt(d_k)


assert d_k == Q.shape[1], "Incorrect d_k"
assert S_scaled.shape == (3, 3), "Incorrect scaled shape"

expected = torch.tensor([
    [1.0 / math.sqrt(2), 1.0 / math.sqrt(2), 2.0 / math.sqrt(2)],
    [0.0 / math.sqrt(2), 1.0 / math.sqrt(2), 1.0 / math.sqrt(2)],
    [1.0 / math.sqrt(2), 2.0 / math.sqrt(2), 3.0 / math.sqrt(2)]
])
assert torch.allclose(S_scaled, expected), "S_scaled không khớp với kết quả tính tay lý thuyết!"



print("Q.shape:", Q.shape)
print("Q.shape[0]:", Q.shape[0])
print("Q.shape[1]:", Q.shape[1])
print("d_k:", d_k)
print("scale:", math.sqrt(d_k))
print("S_scaled:", S_scaled)
  

```
